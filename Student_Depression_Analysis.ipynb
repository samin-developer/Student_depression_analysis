```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.utils import resample
from collections import Counter
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Step 1: Generate synthetic imbalanced dataset
X, y = make_classification(n_samples=1000, n_features=20,
                           n_informative=10, n_redundant=5,
                           n_classes=2, class_sep=2,
                           weights=[0.9, 0.1], random_state=42)

# Step 2: Create DataFrame
feature_names = [f"feature_{i}" for i in range(X.shape[1])]
df_imbalanced = pd.DataFrame(X, columns=feature_names)
df_imbalanced["target"] = y

# Step 3: Split into train and test
X = df_imbalanced.drop(columns="target")
y = df_imbalanced["target"]
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)

# Step 4: Separate classes
X_majority = X_train[y_train == 0]
X_minority = X_train[y_train == 1]
y_majority = y_train[y_train == 0]
y_minority = y_train[y_train == 1]

# Step 5: Oversample minority class
X_minority_oversampled, y_minority_oversampled = resample(
    X_minority, y_minority,
    replace=True,
    n_samples=len(X_majority),
    random_state=42
)

# Step 6: Combine majority with oversampled minority
X_train_oversampled = pd.concat([X_majority, X_minority_oversampled])
y_train_oversampled = pd.concat([y_majority, y_minority_oversampled])

# Step 7: Print class distributions
print("Original class distribution:", Counter(y_train))
print("Resampled class distribution:", Counter(y_train_oversampled))

# Optional: Visualize the new balanced distribution
plt.figure(figsize=(6, 4))
sns.countplot(x=y_train_oversampled, palette="Set1")
plt.title("Class Distribution After Oversampling")
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

```

```python
### 2. **Over-Sampling**

##Over-sampling involves increasing the number of instances in the minority class by duplicating existing samples. This can be done by simply repeating data points, but this may lead to overfitting, where the model learns to memorize the minority class instances instead of generalizing from them.

#### Pros:
## Helps in balancing the dataset.
##Can improve model performance for the minority class.

#### Cons:
##Potential risk of overfitting.
##Does not introduce new data, merely replicates existing data.
```
